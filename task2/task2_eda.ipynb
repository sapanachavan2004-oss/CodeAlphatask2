```python
import os
import pandas as pd

# Settings - change these for your own dataset
FILE = "reviews.csv"
TEXT_COL = "review"
RATING_COL = "rating"

df = pd.read_csv(FILE)
df.head()
```

```python
print("Rows, columns:", df.shape)
print()
print(df.dtypes)
print()
df.info()
```

```python
df.describe(include="all")
```

```python
print("Rating counts:")
print(df[RATING_COL].value_counts().sort_index())
print("Average rating:", round(df[RATING_COL].mean(), 2))
```

```python
# New column: review length in words
df["review_length"] = df[TEXT_COL].astype(str).str.split().str.len()

# Outliers using the IQR rule
q1, q3 = df["review_length"].quantile([0.25, 0.75])
iqr = q3 - q1
low, high = q1 - 1.5 * iqr, q3 + 1.5 * iqr
outliers = df[(df["review_length"] < low) | (df["review_length"] > high)]
print("Outlier reviews (very short or very long):", len(outliers))
outliers
```

```python
df["review_length"] = df[TEXT_COL].astype(str).str.split().str.len()

print(df.groupby(RATING_COL)["review_length"].mean().round(1))
corr = df[RATING_COL].corr(df["review_length"])
print()
print(f"Correlation (rating vs length): {corr:.2f}")
print("Close to 0 = no relationship; close to +1 or -1 = strong relationship")
```

```python
print("Missing values:")
print(df.isnull().sum())
print()
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Clean the data for the next tasks
df = df.drop_duplicates().dropna(subset=[TEXT_COL])
df.to_csv("cleaned_reviews.csv", index=False)
print("Saved cleaned_reviews.csv")
```
